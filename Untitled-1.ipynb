```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


df=pd.read_csv("Used_Car_Price_Prediction.csv")

columns =[          #LIST
    "make",
    "model",
    "variant",
    "yr_mfr",
    "fuel_type",
    "kms_run",
    "city",
    "body_type",
    "transmission",
    "total_owners",
    "sale_price",
    "ad_created_on"
]

smartdeal_df=df[columns].copy()
print(smartdeal_df)
```

```python
smartdeal_df.drop_duplicates(inplace=True)
print(smartdeal_df["ad_created_on"])
```

```python
smartdeal_df["ad_created_on"]=pd.to_datetime(
    smartdeal_df["ad_created_on"],
    errors="coerce")
print(smartdeal_df["ad_created_on"])
```

```python
smartdeal_df.dropna(
    subset="ad_created_on",
    inplace=True
)
```

```python
smartdeal_df=smartdeal_df[smartdeal_df["sale_price"]>5000].copy()
```

```python
features=[
    "make",
    "model",
    "variant",
    "fuel_type",
    "city",
    "body_type",
    "transmission",
    "vehicle_age",
    "kms_run",
    "total_owners"
]
X = smartdeal_df[features]
y = smartdeal_df["sale_price"]

print(X.shape)
print(y.shape)
X.head()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train,y_test=train_test_split(
    X,
    y,
    test_size=0.2, #20%
    random_state=42
)
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
categorical_features = [
    "make",
    "model",
    "variant",
    "fuel_type",
    "city",
    "body_type",
    "transmission"
]

numeric_features = [
    "vehicle_age",
    "kms_run",
    "total_owners"
]

print("Categorical:", categorical_features)
print("Numeric:", numeric_features)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder,StandardScaler

preprocessor= ColumnTransformer(
    transformers=[(
        "categorical",              #could be named anything
        OneHotEncoder(handle_unknown="ignore"),      #If you see a category in test/new data that wasn't present during training, don't crash. Just handle it safely
        categorical_features        #" , "
    ),
    (
        "numeric",
        StandardScaler(),
        numeric_features
    )]
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression

linear_model=Pipeline(
    steps=[
        ("preprocessor",preprocessor),
        ("model",LinearRegression())
    ]
)
```

```python
linear_model.fit(X_train,y_train)
```

```python
y_predict= linear_model.predict(X_test)
print(y_predict[:10])
```

```python
results=pd.DataFrame({
    "Acutal Price": y_test.values,
    "Predicted Price": y_predict
})
results.head(10)
```

```python
results.rename(
    columns={"Acutal Price": "Actual Price"},
    inplace=True
)
```

```python
results["Difference"] = (
    results["Predicted Price"]
    - results["Actual Price"]
)

results.head(10)
```

```python
results["Absolute Error"] = abs(
    results["Difference"]
)

results.head(10)
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score
from sklearn.metrics import mean_squared_error
import numpy as np

# MEAN ABOLUTE ERROR
mae = mean_absolute_error(y_test,y_predict)


# R SQUARE
r2s = r2_score(y_test,y_predict)


# ROOT MEAN SQUARED ERROR
rmse=np.sqrt(mean_squared_error(y_test,y_predict))


print("Mean Absolute Error:", mae)
print("Root Mean Squared Error:", rmse)
print("R2 Score:", r2s)
```

```python
analysis_df = X_test.reset_index(drop=True).copy()      #New DF

analysis_df["Actual Price"] = y_test.reset_index(drop=True)
analysis_df["Predicted Price"] = y_predict

analysis_df["Absolute Error"]=abs(analysis_df["Actual Price"]-analysis_df["Predicted Price"])       #DIFF

worst_prediction = analysis_df.sort_values(
    by="Absolute Error",
    ascending=False
)
worst_prediction.head(10)
```

```python
from sklearn.ensemble import RandomForestRegressor

random_forest = Pipeline(
    steps=[
        ("preprocessor",preprocessor),
        ("model", RandomForestRegressor(
            n_estimators=200,               # NUMBER OF TREES TO PREDICT 
            random_state=42,                
            n_jobs=-1                       # RandomForestRegressorThis controls how many CPU cores sklearn can use. -1 IS TAKE ALL THE FREE CORE
        ))
    ]
)
```

```python
random_forest.fit(X_train,y_train)
```

```python
rf_predict= random_forest.predict(X_test)
```

```python
rf_mae = mean_absolute_error(y_test, rf_predict)
rf_rmse=np.sqrt(mean_squared_error(y_test, rf_predict))
rf_r2=r2_score(y_test,rf_predict)

print("Random Forest MAE:", rf_mae)
print("Random Forest RMSE:", rf_rmse)
print("Random Forest R2:", rf_r2)
```

```python
rf_results=pd.DataFrame({
    "Actual Price" : y_test,
    "Predicted Price" : rf_predict
})

rf_results["Absolute Error"]=abs(rf_results["Actual Price"]-rf_results["Predicted Price"])

rf_results["Percentange Error"]=(rf_results["Absolute Error"]/rf_results["Actual Price"])*100   #(|Actual-predictaed|/Actual)*100

rf_mape=rf_results["Percentange Error"].mean()

rf_results.head(10)
print("MAPE",rf_mape)
```

```python
def get_deal(predicted_price, asking_price, tolerance):
    
    difference=asking_price-predicted_price
    diff_pecent=(difference/predicted_price)*100

    if diff_pecent < -(2*tolerance):
        verdict="Excellent Deal"

    elif diff_pecent < -tolerance:
        verdict="Good Deal"

    elif diff_pecent <= tolerance:
        verdict="Fair Deal"

    elif diff_pecent< (2*tolerance):
        verdict="Overpriced"

    else:
        verdict="Very Overpriced"

    return difference,diff_pecent,verdict
```

```python
def smartdeal_predict(
    make,
    model,
    variant,
    fuel_type,
    city,
    body_type,
    transmission,
    vehicle_age,
    kms_run,
    total_owners,
    asking_price
):

    car = pd.DataFrame([{
        "make": make,
        "model": model,
        "variant": variant,
        "fuel_type": fuel_type,
        "city": city,
        "body_type": body_type,
        "transmission": transmission,
        "vehicle_age": vehicle_age,
        "kms_run": kms_run,
        "total_owners": total_owners
    }])

    predicted_price = random_forest.predict(car)[0]

    difference, diff_percent, verdict = get_deal(
        predicted_price,
        asking_price,
        rf_mape
    )

    print("----- SMARTDEAL REPORT -----\n")
   
    print("Estimated Fair Price: ₹", round(predicted_price, 2))
    print("Seller Asking Price: ₹", round(asking_price, 2))
    print("Price Difference: ₹", round(difference, 2))
    print("Percentage Difference:", round(diff_percent, 2), "%")
    print()

    print("Verdict:", verdict)
```
