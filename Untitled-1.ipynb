```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


df=pd.read_csv("Used_Car_Price_Prediction.csv")

columns =[          #LIST
    "make",
    "model",
    "variant",
    "yr_mfr",
    "fuel_type",
    "kms_run",
    "city",
    "body_type",
    "transmission",
    "total_owners",
    "sale_price",
    "ad_created_on"
]

smartdeal_df=df[columns].copy()
print(smartdeal_df)
```

```python
smartdeal_df.drop_duplicates(inplace=True)
print(smartdeal_df["ad_created_on"])
```

```python
smartdeal_df["ad_created_on"]=pd.to_datetime(
    smartdeal_df["ad_created_on"],
    errors="coerce")
print(smartdeal_df["ad_created_on"])
```

```python
smartdeal_df.dropna(
    subset="ad_created_on",
    inplace=True
)
```

```python
smartdeal_df=smartdeal_df[smartdeal_df["sale_price"]>5000].copy()
```

```python
type(smartdeal_df)
```

```python
smartdeal_df["body_type"] = (
    smartdeal_df["body_type"]
    .fillna("unknown")
)

smartdeal_df["transmission"] = (
    smartdeal_df["transmission"]
    .fillna("unknown")
)
```

```python
smartdeal_df["listing_year"]=(
    smartdeal_df["ad_created_on"].dt.year
)
smartdeal_df["vehicle_age"]=(
    smartdeal_df["listing_year"]-
    smartdeal_df["yr_mfr"]
)


```

```python
smartdeal_df.drop(
    columns=["ad_created_on"],
    inplace=True
)
```

```python
print("Shape:",smartdeal_df.shape)
print("\nmissing val:",smartdeal_df.isnull().sum())
print("\nduplicate:", smartdeal_df.duplicated().sum())

print("\nNumerical Summary:")
print(
    smartdeal_df[
        [
            "vehicle_age",
            "kms_run",
            "total_owners",
            "sale_price"
        ]
    ].describe()
)
```

```python
features=[
    "make",
    "model",
    "variant",
    "fuel_type",
    "city",
    "body_type",
    "transmission",
    "vehicle_age",
    "kms_run",
    "total_owners"
]
X = smartdeal_df[features]
y = smartdeal_df["sale_price"]

print(X.shape)
print(y.shape)
X.head()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train,y_test=train_test_split(
    X,
    y,
    test_size=0.2, #20%
    random_state=42
)
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
categorical_features = [
    "make",
    "model",
    "variant",
    "fuel_type",
    "city",
    "body_type",
    "transmission"
]

numeric_features = [
    "vehicle_age",
    "kms_run",
    "total_owners"
]

print("Categorical:", categorical_features)
print("Numeric:", numeric_features)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder,StandardScaler

preprocessor= ColumnTransformer(
    transformers=[(
        "categorical",              #could be named anything
        OneHotEncoder(handle_unknown="ignore"),      #If you see a category in test/new data that wasn't present during training, don't crash. Just handle it safely
        categorical_features        #" , "
    ),
    (
        "numeric",
        StandardScaler(),
        numeric_features
    )]
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression

linear_model=Pipeline(
    steps=[
        ("preprocessor",preprocessor),
        ("model",LinearRegression())
    ]
)
```

```python
linear_model.fit(X_train,y_train)
```

```python
y_predict= linear_model.predict(X_test)
print(y_predict[:10])
```

```python
results=pd.DataFrame({
    "Acutal Price": y_test.values,
    "Predicted Price": y_predict
})
results.head(10)
```

```python
results.rename(
    columns={"Acutal Price": "Actual Price"},
    inplace=True
)
```

```python
results["Difference"] = (
    results["Predicted Price"]
    - results["Actual Price"]
)

results.head(10)
```

```python
results["Absolute Error"] = abs(
    results["Difference"]
)

results.head(10)
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score
from sklearn.metrics import mean_squared_error
import numpy as np

# MEAN ABOLUTE ERROR
mae = mean_absolute_error(y_test,y_predict)


# R SQUARE
r2s = r2_score(y_test,y_predict)


# ROOT MEAN SQUARED ERROR
rmse=np.sqrt(mean_squared_error(y_test,y_predict))


print("Mean Absolute Error:", mae)
print("Root Mean Squared Error:", rmse)
print("R2 Score:", r2s)
```
